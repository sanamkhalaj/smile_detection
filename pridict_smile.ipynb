```python
from keras.models import load_model
import cv2 as cv
import glob
import numpy as np
```

```python
model=load_model(r"c:\Users\user\Desktop\smilehoosh\smile_model.keras")
```

```python
adress=r"C:\Users\user\Desktop\archive 2"
```

```python
for folder in glob.glob(adress+"\\*"):
    for img_p in glob.glob(folder+"\\*"):
        img=cv.imread(img_p)
        if img is None:
            continue
        img=cv.cvtColor(img,cv.COLOR_BGR2RGB)
        img=img.flatten()
        img=img/255.0
        img=np.expand_dims(img,axis=0)
        out=model.predict(img)
        if out[0][0]>0.5:
            print(img_p,"smile")
        else:
            print(img_p,"nosmile")
```
